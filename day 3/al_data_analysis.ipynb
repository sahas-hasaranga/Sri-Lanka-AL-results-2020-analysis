First, let's clean the `district_rank` column by extracting only the numerical part and converting it to an integer. Then we can filter for ranks less than 500.

```python
# Function to extract the numerical part from 'district_rank'
def extract_rank(rank_str):
    if isinstance(rank_str, str):
        try:
            return int(rank_str.split(' ')[0])
        except ValueError:  # Catch error if conversion to int fails (e.g., for '-')
            return None
    return None

# Apply the function to create a new numerical rank column
df['district_rank_numeric'] = df['district_rank'].apply(extract_rank)

# Filter the DataFrame for district_rank_numeric less than 500
df_filtered_rank = df[df['district_rank_numeric'] < 500]

print("Shape of the filtered DataFrame:", df_filtered_rank.shape)
print("\nHead of the filtered DataFrame (district_rank < 500):")
display(df_filtered_rank.head())
```

```python
print('Before dropping nulls and duplicates:')
print('Shape:', df.shape)
print('Missing values in gender:', df['gender'].isnull().sum())
print('Duplicate rows:', df.duplicated().sum())

# Drop rows where 'gender' is null
df_cleaned = df.dropna(subset=['gender']).copy()

# Drop duplicate rows from the entire DataFrame
df_cleaned = df_cleaned.drop_duplicates().copy()

print('\nAfter dropping nulls and duplicates:')
print('Shape:', df_cleaned.shape)
print('Missing values in gender:', df_cleaned['gender'].isnull().sum())
print('Duplicate rows:', df_cleaned.duplicated().sum())

display(df_cleaned.head())
```

```python
import pandas as pd

def extract_rank(rank_str):
    """Extracts the numerical part from the district_rank string."""
    if isinstance(rank_str, str):
        try:
            return int(rank_str.split(' ')[0])
        except ValueError:
            return None
    return None

def transform_data(input_file_path, output_file_path):
    print(f"Loading data from {input_file_path}...")
    df = pd.read_csv(input_file_path)

    print("Initial Data Shape:", df.shape)

    # 1. Drop missing values in specific columns (e.g., gender)
    print("Dropping rows with missing gender...")
    df_cleaned = df.dropna(subset=['gender']).copy()

    # 2. Drop duplicate rows
    print("Dropping duplicate rows...")
    df_cleaned = df_cleaned.drop_duplicates().copy()

    # 3. Extract numerical district rank
    print("Extracting numerical district ranks...")
    df_cleaned['district_rank_numeric'] = df_cleaned['district_rank'].apply(extract_rank)

    print("Final Data Shape after transformations:", df_cleaned.shape)

    # Save the transformed dataset
    print(f"Saving transformed data to {output_file_path}...")
    df_cleaned.to_csv(output_file_path, index=False)
    print("Transformation complete! 🚀")

if __name__ == "__main__":
    # Define file paths (Change these if running in Colab/Drive)
    INPUT_FILE = "/content/2020_al_data_kaggle_upload_new_old_syllabi.csv"
    OUTPUT_FILE = "cleaned_al_data.csv"

    transform_data(INPUT_FILE, OUTPUT_FILE)
```
